#### Task 4 **[30 points]**

Write some python program that generates a plot showing the monthly mean temperatures for Germany and a second country of your choice in the year 2024. Make sure to briefly describe/interpret what you can see in the plot.

_Hint 1: [Matplotlib](https://matplotlib.org) and [seaborn](https://seaborn.pydata.org) are valuable resources for plotting data from pandas dataframes._<br>
_Hint 2: Label your axes, and add a concise title. Make sure every part of the plot is human-readable. There are plenty of tutorials about how to visualize data with python available online!_


```python
seasons = ['DJF', 'MAM', 'JJA', 'SON']
temp_cols = [col for col in df.columns if 'TMP_MEAN_' in col and not any(s in col for s in seasons)]
# temp_cols
df_ = df.query('YEAR == 2024 and (COUNTRY == "GERMANY" or COUNTRY == "IRAN")')
df_ = df_[['COUNTRY', 'YEAR'] + temp_cols]
df_
```

```python
months_pretty = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']

plt.figure(figsize=(10,6))
for _, row in df_.iterrows():
    plt.plot(months_pretty, row[temp_cols], marker='o', label=row['COUNTRY'])

plt.title('Monthly Mean Temperatures for Germany and Iran in 2024')
plt.xlabel('Month')
plt.ylabel('Mean Temperature')
plt.legend()
plt.grid(True)
plt.show()
```
