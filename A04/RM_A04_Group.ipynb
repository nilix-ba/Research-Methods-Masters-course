##### **Task 1.7:** Define and assess the validity and reliability of your experiment and its measurements (a written explanation without calculations is sufficient).



###### **Section 1 – Demographics and location**

**Validity**

* Age and gender have very high **face validity**: they clearly measure what they claim.
* “Type of location” directly represents our construct for Research Question B (living environment).
* Social media hours are a reasonable **proxy** for social media intensity, which is a relevant control variable.

**Reliability**

* Age, gender, and location type are **stable characteristics**, so answers should be consistent.
* Social media hours are self-estimates, but broad categories (e.g. 1–2h, 2–3h) help keep responses **reasonably consistent** across time and people.

---

###### **Section 2 – Exposure to AI-generated media (IV1)**

**Validity**

* The three items together cover different aspects of AI exposure: seeing AI images/videos, reading AI text, and using AI tools.
* Examples like “AI art, deepfakes, ChatGPT” make it clear what counts as AI-generated content, which improves **construct validity**.

**Reliability**

* All items use the same 1–5 frequency scale, which supports consistent answering.
* Combining items 5–7 into one **mean AI exposure score** increases reliability compared to a single question (higher internal consistency).

---

###### **Section 3 – Trust in online content (DV1)**

**Validity**

* Items 8 and 9 measure **trust** directly (trust and reliability), while item 10 taps into **skepticism** as the opposite side.
* Together, they represent the construct “trust in online content” better than only one simple item.

**Reliability**

* A 3-item Likert scale with a shared 1–5 response format is suitable for a **reliable composite score**.
* The reverse-coded item (10) helps detect response patterns and can improve the internal consistency of the scale.

---

###### **Section 4 – Reduction in trust in social media content (DV2)**

**Validity**

* All three items explicitly ask about **change over time** (compared to the past, has gone down), which matches our construct “reduction in trust”.
* This makes the scale very well aligned with your Research Question B.

**Reliability**

* Because items 11–13 all point in the same direction (higher = more reduction), they can form a **coherent scale**.
* Similar wording and the same 1–5 format support consistent responses and good internal reliability.

---

###### **Section 5 – Misinformation (control)**

**Validity**

* The question directly asks how often participants later realized something was **false or misleading**, which fits well as a measure of “experience with misinformation”.

**Reliability**

* It is a **single-item** measure, so its reliability is more limited than a multi-item scale.
* Still, the question is simple, and the 1–5 frequency scale is clear, so responses should be sufficiently consistent for use as a control variable.

---

#### **Study design & data handling (from the “How we would conduct the experiment” part)**

**Validity**

* The **online survey** format fits our topic (internet and social media use) and reaches the right target group.

* $H_0$: *After controlling for X and W, C has no direct effect on Y.*
> Y ~ X + W

* $H_1$: *C has a direct effect on Y even after X and W.*
> Y ~ X + W + C

Assuming Y is the output and (W, X, C) are the inputs, we would like to determine whether or not the model can be simplified to not needing C or whether it is an integral part of the model and is required to determine Y.

#### Experimental Procedure

To determine which model is more fitting to our data, simplified or not, we will run:
* nested linear regression
* Anova test (F-test)

We will be conducting these tests as we have a simplified model for out $H_0$ and can easily compare the results with the F-test.

### Task 3.4

```python
import pandas as pd
import statsmodels.formula.api as smf
import statsmodels.api as sm
```

```python
exp_df = pd.read_csv("/content/drive/MyDrive/RM seminar/A04/data/experiment_data.csv")
# exp_df.head()
```

```python
model_reduced = smf.ols("Y ~ X + W", data=exp_df).fit()

model_full = smf.ols("Y ~ X + W + C", data=exp_df).fit()

print('Simplified Model: Y ~ X + W', "\n")
print(model_reduced.summary(), '\n')
print(f"Actual p-value for X: {model_reduced.pvalues['X']:.10e}")
print(f"Actual p-value for W: {model_reduced.pvalues['W']:.10e}", '\n')

print('Entire Model: Y ~ X + W + C', '\n')
print(model_full.summary(), '\n')
print(f"Actual p-value for C[T.B]: {model_full.pvalues["C[T.B]"]:.10e}")
print(f"Actual p-value for C[T.C]: {model_full.pvalues["C[T.C]"]:.10e}")
```

Here we see the results of the simplified model ($M_0$) as well as that of the full model ($M_1$).

For $M_0$, we can tell through the R-squared value of 0.752, that roughly 75.2% of the variance in Y is explained by the linear combination of X and W. This is relatively high and allows us to assume that a linear model was the correct fit. When moving on to the X and W coefficients, when W is constant and X is changed by one, on average Y will increase by 0.41 (std. error = 0.017). X also has a t-value of 23.85 (p = 2.22 × 10⁻⁸⁴). All of this shows a relationship between X and Y to be significant and X is a strong predictor of Y.

The W coefficiant, similarily to X, also increases Y. When changing W by 1 and X remains constant, Y changes by roughly 1.00 (std. error = 0.032). Thus, showing a parent relationship between the two variables. The t-value rounded is 31.30 with an even smaller p-value of 1.39 × 10⁻¹¹⁹. From this we can denote that W has a strong linear relationship with Y.

Now inspecting $M_1$, we can see that the R-squared value is essentially the same as it was for $M_0$. From the C variable we can see negative coefficients, C[T.B] = -0.04 (std. error = 0.179) and C[T.C] = -0.05 (std. error = 0.183). The t-values are similar between the two, while also being negative values, C[T.B] t = -0.223 (p = 0.824) and C[T.C] = -0.277 (p = 0.782). With these small t-values and high p-values, it can be assumed that give X and W there is no detectable difference in Y between A vs B and A vs C. In addition, the X and W coeficients remain the similar to the $M_0$ results. Therefore, adding C as an extra parameter didn't help in predicting Y's value.

```python
print('Anova Test', '\n')
anova_res = sm.stats.anova_lm(model_reduced, model_full)
print(anova_res)
```

In this Anova test, the residual sum of squares for $M_0$ is 1273.66 and that of $M_1$ is 1273.41 with a difference of 0.249. Adding C to the model $M_1$ fractionally reduced the SSR. Thus also leading to the difference in degrees of freedom to account for the additional parameters of C[T.B] and C[T.C].

The F-statistic is 0.048 is relatively low with a very high p value of 0.952. Due to the F-statistic being so small and the p-value so high, we can say that the difference in SSRs is very likely due to chance.

```python
r2_reduced = model_reduced.rsquared
r2_full    = model_full.rsquared
delta_r2   = r2_full - r2_reduced

print("R2 reduced:", r2_reduced)
print("R2 full   :", r2_full)
print("Delta R2  :", delta_r2)
```

In this study, we try to reveal the underlying model of the Magic Data API, taking the latter as an observational study and analyzing together the behavior of three continuous variables, X, W, and Y, along with a categorical label C.

At first, category C seemed very important: points labeled A or B seemed to have much higher Y values than those labeled C. Closer inspection showed that C was mainly tracking W — negative W values were mostly labeled C, and positive values were mostly labeled A or B — but that C added almost nothing once X and W were known. More formally, we compared two models: one including only X and W, the other including X, W and C. The R² for the model including C was less than 0.01% higher than that of the model without C. A nested F-test found no evidence that C had any direct effect given inclusion of X and W in the model. These results taken together support the hypothesis that the API first draws X and W, then generates Y as a stable linear function of them, and only afterwards assigns C as a label derived mainly from W. This implies that C does not have an independent causal role in determining Y, thus we fail to reject our $H_0$.
